```python
bio = pd.read_excel('./input/28072025/Bio.xlsx')
events = pd.read_excel('./input/28072025/Events.xlsx')
spouses = pd.read_excel('./input/28072025/Echtgenoten.xlsx')
```

```python
# Panda settings for showing data (this is foremost done to more easily explore the data while processing it)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
# convert geboortjaar_int and sterfjar_int to integers

bio['geboortejaar_int'] = bio['geboortejaar_int'].fillna(0).astype('int')
bio['geboortejaar_int'] = bio['geboortejaar_int'].astype('int')
bio['sterfjaar_int'] = bio['sterfjaar_int'].fillna(0).astype('int')
bio['sterfjaar_int'] = bio['sterfjaar_int'].astype('int')
```

```python
bio.rename(columns={
    'id':'id',
    'original_input':'original_input', #remove
    'titel':'titel',
    'achternaam':'achternaam',
    'tussenvoegsel':'tussenvoegsel',
    'voornamen of voorletters':'voornamen_of_voorletters',
    'beroep(en)':'beroep_en',
    'titulatuur (geslacht)':'titulatuur_geslacht',
    'geslacht':'geslacht',
    'oorsprong':'oorsprong',
    'verwijzing':'verwijzing',
    'verwzijzing_id':'verwzijzing_id',
    'geb/overl.':'geb/overl.', #remove
    'geboortejaar':'geboortejaar',
    'geboortejaar_int':'geboortejaar_int',
    'sterfjaar':'sterfjaar',
    'sterfjaar_int':'sterfjaar_int',
    'bron, etc, >naar events of spouses':'bron, etc, >naar events of spouses', #remove
    'Overige details':'overige details',
    'opmerkingen_wildcard':'opmerkingen_wildcard', #remove
    'CHECKED (1= no adjustments, 2=manual adjustments])':'CHECKED'#remove
}, inplace=True)

bio['vde_dz']= "ZDAB"
```

```python
bio = bio.drop(['original_input', 
                'geb/overl.', 
                'bron, etc, >naar events of spouses',
                'opmerkingen_wildcard',
                'CHECKED'
                ], axis=1)
```

```python
events.rename(columns={
    'id':'id',
    'werkgebied en -soort_org':'werkgebied en -soort_org', #remove
    'bijzonderheden_org':'bijzonderheden_org', #remove
    'organ.':'organisatie',
    'werkgebied en -soort':'werkgebied_en_soort',
    'werkperiode':'werkperiode', #remove
    'periode_start':'periode_start',
    'periode_start_int':'periode_start_int',
    'periode_einde':'periode_einde',
    'periode_einde_int':'periode_einde_int',
    'bijzonderheden/opmerkingen':'bijzonderheden_opmerkingen',
    'event':'event',
    'bio_info':'bio_info',
    'bronvermelding':'bronvermelding',
    'details overlijden':'details_overlijden',
    'vervolgrol(len)':'vervolgrol_en_nevenrollen',
    'info_spouse':'info_spouse', #remove
    'info toegevoegd uit bio':'info toegevoegd uit bio', #remove
    'welke info':'welke info', #remove
    'wildcard':'wildcard', #remove
    'opmerkingen_wildcard':'opmerkingen_wildcard', #remove
    'CHECKED (1= no adjustments, 2=manual adjustments])':'CHECKED'#remove
}, inplace=True)
```

```python
events = events.drop(['werkgebied en -soort_org',
                       'bijzonderheden_org',
                       'werkperiode',
                       'info_spouse',
                       'info toegevoegd uit bio',
                       'welke info',
                       'wildcard',
                       'opmerkingen_wildcard',
                       'CHECKED'
                       ], axis=1)
```

```python
spouses.rename(columns={
    'spouse':'spouse', #remove
    'ID':'id',
    'Naam echtgenoot':'naam',
    'Geboortejaar':'geboortejaar',
    'Sterfjaar':'sterfjaar',
    'Trouwjaar':'trouwjaar',
    'Opmerkingen':'opmerkingen',
    'Eind huwelijk':'eind_huwelijk'
}, inplace=True)
```

```python
spouses = spouses.drop(['spouse'], axis=1)
```

```python
#update id fields into a string so they remain distinguishable from the other lists

bio['id'] = bio['id'].apply(lambda x: f"z_{x:04d}")
events['id'] = events['id'].apply(lambda x: f"z_{x:04d}")
spouses['id'] = spouses['id'].apply(lambda x: f"z_{x:04d}")
```

```python
date_str = datetime.today().strftime('%m_%d_%Y')

filename_bio = "output//"+ f'bio_{date_str}.xlsx'
bio.to_excel(filename_bio, index=False)

filename_events = "output//"+ f'events_{date_str}.xlsx'
events.to_excel(filename_events, index=False)

filename_spouses = "output//"+ f'echtgenoten_{date_str}.xlsx'
spouses.to_excel(filename_spouses, index=False)
```
